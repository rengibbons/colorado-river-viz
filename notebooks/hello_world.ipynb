# Hello World

This notebook demonstrates importing from the `myproject` package and doing basic analysis.

```python
import matplotlib.pyplot as plt

from myproject import hello_world

%matplotlib inline
```

```python
data = hello_world()

print(f"Shape:  {data.shape}")
print(f"Mean:   {data.mean():.4f}")
print(f"Std:    {data.std():.4f}")
print(f"Min:    {data.min():.4f}")
print(f"Max:    {data.max():.4f}")
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(data, bins=20, edgecolor="white")
ax.set_title("Sample data from myproject.hello_world()")
ax.set_xlabel("Value")
ax.set_ylabel("Count")
plt.tight_layout()
plt.show()
```
